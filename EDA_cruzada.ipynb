```python
import os
import pandas as pd
import matplotlib
matplotlib.use('Agg') # Essencial para não sobrecarregar a memória
import matplotlib.pyplot as plt
from analise_cruzada import carregar_e_preparar_todos, plotar_series_sobrepostas, plotar_correlacao_cruzada

# 1. Definição dos Caminhos
caminhos_dict = {
    "CCET": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_CCET.csv",
    "BIO": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_BIO.csv",
    "CCH": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_CCH.csv",
    "CCSO": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_CCSO.csv",
    "IEE": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_IEE.csv",
    "ODONTO": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_ODONTO.csv",
    "PF": "/home/ls/TCC_2026/Dados greefra/mes_1/dados_limpos/Medidor_PF.csv"
}

# 2. Carrega e junta todos os dados em uma única matriz
df_campus = carregar_e_preparar_todos(caminhos_dict)

# 3. Loop para gerar os gráficos cruzados mês a mês
for ano in [2025, 2026]:
    for mes in range(1, 13):
        # Evita processar meses que não existem nos dados
        if df_campus.index.min() <= pd.Timestamp(f"{ano}-{mes:02d}-01") <= df_campus.index.max():
            
            # Gera o gráfico de linhas sobrepostas (para ver as quedas simultâneas)
            plotar_series_sobrepostas(df_campus, ano, mes)
            
            # Gera a matriz de correlação (para ver a dependência entre prédios)
            plotar_correlacao_cruzada(df_campus, ano, mes)
```
